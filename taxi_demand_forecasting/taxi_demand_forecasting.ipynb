```python
def make_features(df, max_lag, rolling_mean_size):
    df = df.copy()
    df['month'] = df.index.month
    df['day'] = df.index.day
    df['dayofweek'] = df.index.dayofweek
    df['hour'] = df.index.hour
    for lag in range(1, max_lag + 1):
        df[f'lag_{lag}'] = df['num_orders'].shift(lag)
    df['rolling_mean'] = df['num_orders'].shift().rolling(rolling_mean_size).mean()
    return df
```

```python
plt.figure(figsize=(15, 6))
df['num_orders'].plot()
plt.title('Hourly Taxi Orders Over Time')
plt.show()
```

The number of orders changes over time and shows repeated fluctuations, suggesting the presence of temporal patterns.

```python
df['hour'] = df.index.hour
hourly_avg = df.groupby('hour')['num_orders'].mean()

plt.figure(figsize=(12, 5))
hourly_avg.plot(kind='bar')
plt.title('Average Taxi Orders by Hour')
plt.show()
```

Average demand differs by hour, so the hour feature is likely useful for prediction.

```python
df['dayofweek'] = df.index.dayofweek
daily_avg = df.groupby('dayofweek')['num_orders'].mean()

plt.figure(figsize=(10, 5))
daily_avg.plot(kind='bar')
plt.title('Average Taxi Orders by Day of Week')
plt.show()
```

Demand also varies by day of week, which supports using the dayofweek feature.

Before creating new features, I performed exploratory data analysis to identify temporal patterns in the series. The plots showed that taxi demand changes by hour and day of week and also depends on recent previous values, which justified the use of calendar, lag, and rolling mean features.

```python
df = make_features(df, max_lag=24, rolling_mean_size=24)
df.dropna(inplace=True)
```

## Model Training

In this section, I will split the dataset into training and test sets, then train several models and compare their performance.

```python
train_size = int(len(df) * 0.9)

train = df.iloc[:train_size]
test = df.iloc[train_size:]

X_train = train.drop('num_orders', axis=1)
y_train = train['num_orders']

X_test = test.drop('num_orders', axis=1)
y_test = test['num_orders']

print(X_train.shape)
print(X_test.shape)
```

For time series, the train-test split must preserve the chronological order of observations. The model is trained on earlier data and tested on later data.

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_train)
lr_rmse = np.sqrt(mean_squared_error(y_train, lr_pred))

lr_rmse
```

The Linear Regression model was trained on the training set, and the RMSE on the training set was calculated. The test set is reserved for the final evaluation step only.

Next, I will tune the Random Forest model to see whether a better combination of hyperparameters can improve the result.

```python
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.ensemble import RandomForestRegressor

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None]
}

tscv = TimeSeriesSplit(n_splits=3)

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=12345),
    param_grid=param_grid,
    cv=tscv,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_
grid_search.best_params_
```
